```python
from pyspark.sql import functions as f
```

```python
# Data transformation
df_bronze=spark.table("cockpit.finance.bronze_invoice")
display(df_bronze)
```

```python
# SLA Breached transformation 1= breached, 0= not breached
df_silver_sla_breached=df_bronze.withColumn("SLA_Breached", f.when(f.col("Hours")>30,1).otherwise(0))
display(df_silver_sla_breached)
```

```python
df_silver_sla_breached.write.mode("overwrite").saveAsTable("cockpit.finance.silver_invoice")
```
